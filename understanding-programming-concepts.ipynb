### Task 1: Word Frequency Identifying Algorithm

This task was about creating an algorithm to find the most commonly used word in a given text. This example comes from pages 20-22 in the Course Lab Book and is based on a provided pseudocode. The pseudocode provided states: 

```python
INPUT: poem text
split text into list of words and assign it to a variable called “words”
created an empty dictionary and assign it to a variable called “counts”
FOR each word in words:
    IF word is already in the dictionary counts:
    increase the current count of the word by one
    assign the new value to the entry for the word in the dictionary
ELSE:
    Assign the value 1 to the entry for the word in the dictionary
FIND word with maximum count
OUTPUT that word
```

This pseudocode is basically saying, we input the poem text, tell the code to split the poem into words (defining a variable), insert each word found into a local dictionary, each time a word appears in the dictionary to add one to the count of it, and assign that value to that word. Then, identify the word with the highest number and print that out for us. 

First, we must import libraries so the code has something to pull pre-existing and pre-tested instructions from. Basically, so the coder using the library does not have to build their own functions from scratch every time. 

```python
from collections import Counter
import re

# Okay, excellent. The code libraries are imported. Counter counts items, re is regular expressions for pattern matching in text. 
# Imported libraries contain pre-existing functions, which can be helpful as coders do not have to set everything up themselves every time. Shared libraries also allow a sharing of code between multiple people across systems. 

poem = """You do not have to be good.
You do not have to walk on your knees
For a hundred miles through the desert, repenting.
You only have to let the soft animal of your body
love what it loves.
Tell me about despair, yours, and I will tell you mine.
Meanwhile the world goes on.
Meanwhile the sun and the clear pebbles of the rain
are moving across the landscapes,
over the prairies and the deep trees,
the mountains and the rivers.
Meanwhile the wild geese, high in the clean blue air,
are heading home again.
Whoever you are, no matter how lonely,
the world offers itself to your imagination,
calls to you like the wild geese, harsh and exciting --
over and over announcing your place
in the family of things.""" 

# This tells the code that the term "poem" means this block of text. This is also called "naming a variable" where "poem" is the variable and the contents are the value. 
# The quotation marks make it so the contents are their own unit, as opposed to being broken down and misunderstood as improperly called functions or new, undefined variables. 
# The equal sign is what tells the code to store the contents inside the defined variable. 

words = re.findall(r"\b[\w']+\b", poem.lower())

# This bit is confusing, but basically this function removes the punctuation and makes all the letters lowercase. 
# poem.lower makes it all lowercase. re.findall() calls the function to search the text and return every match of the pattern. 
# r"\b[\w']+\b" is a regular expression. r tells it to interpret the slashes literally. \b is a word boundary. 
# Square brackets contain another function, \w' says to include words and also apostophes, because they're part of compound words. 
# words = names a new function 
# It translates to: find a sequence of one or more word characters or apostrophes that starts and ends at a word boundary. 

counts = Counter(words)
most_common = counts.most_common(1)[0]

# This is a "finding the most common word" function that will work much better now that I have "cleaned" or "normalized" the text. 

print(f"Most frequent word: '{most_common[0]}' appears {most_common[1]} times")

# And this will show us the results. 
```

The final product shows the most frequently appearing word is "the". That is rather anti-climactic. Technically, you could do this with any piece of writing, provided it is input correctly. 

This piece of code is an algorithm, or a series of instructions in a particular order to reach a particular result. The steps are: 
+ Input words for a poem 
+ Split the poem into individual words, rather than one big lump of text 
+ Remove spaces, punctuation, and capitalization 
+ Count how many words there are 
+ Find the word with the highest count 
+ Display those results and report it as the "Most frequent word:" 

The skills shown here are: 
+ Importing code libraries 
+ Naming and calling functions 
+ Understanding of how to order an algorithm 

### Task 2: Pulling an Old Website from the Internet Archive

For this task, I will demonstrate another algorithm (or series of steps in a certain order) to find and display an old website from the Internet Archive. This one requires interaction with the internet and must have certain libraries imported to function properly. This is a useful task for retrieving old versions of websites for research purposes, or just for fun. 

```python
import webbrowser

# This library is an interface for remotely launching and interacting with web browsers like Safari, Chrome, DuckDuckGo, etc. 
# Basically opens URLs in the default web browser of the user. 

import requests 

# This library allows python to send requests to websites and APIs. 

print("We are going to find an old website! How cool!")

# print is a basic function being called to display the text inside the brackets. The quotation marks ensure python does not assume the text is a string of characters and instead its own chunk of data. 
# I have made mine a bit whimsical because this is a long and annoying project and I need to have fun with it. 

# I will define some variables now, to make a shorthand for certain functions I want to run. 

site = input("Type your website URL:") 

# This is a prompt to the user to paste a current website. 

era = input("Type year, month, and day as numbers with no spaces. For example: 20031204.") 

# This is another prompt to the user to input some more information. In this case, a date in a specific way. 

url = "https://archive.org/wayback/available?url=%s&timestamp=%s" % (site,era)

# % is a placeholder for text. 
# This bit will input the site and era into the link and automatically search the Wayback Machine based on teh input data. 

response = requests.get(url) 

# This variable is contacting the Wayback Machine with our URL. 

data = response.json()

# The Wayback Machine responds with JSON data, which we have not learned and thankfully do not need to. 
# The JSON response (in case someone has curosity) will search for the closest date to the input one at the URL given. 

# Now we enter the "try" block. This is what we want python to try and complete, if possible. If this section does not work, there will be a contingency in the next section after the "try". 

try: 
    old_site = data["archived_snapshots"]["closest"]["url"]
    # Here we define a new variable. 
    print("Unearthed this treasure: ", old_site) 
    # This prints a message alongside the newly defined variable. 
    print("It should appear in your browser shortly.")
    webbrowser.open(old_site) 
    # webbrowser will ask your browser to open the link. 

# This section is the contingency if the "try" block fails to work. More accurately, if the Wayback Machine cannot find a website that meets the input date and URL, this is what gets displayed instead. 

except: 
    print("Sorry friend. :( No site for you.") 

# And that should be everything. 

# I am going to run my own test and the below reflection on the task will show the results. 
```
