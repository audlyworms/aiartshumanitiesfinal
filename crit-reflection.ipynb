### My Creative Response

For my proposed LLM, I would like to create a recipe chatbot that can provide "traditional" or regional recipies for places around the world. The chatbot would request a user to input a location and search for a recipe from that region through the internet (but eventually would switch to user-input recipes) alongside a history of the dish. The output would be a map with pins where the recipes are from, a list of those recipes (and the sources of the recipes, e.g. blogs, books, etc.), and a short history of the region's dishes. There should also be an option for users to input their own recipes to share with other users of this LLM. 

The idea is that the LLM will compile a list of dishes from a region and a bit of factual information about them, rather than a user having to go on a Google deep dive to figure out which pupusa is most authentic to Honduras cuisine as well as the easiest to make at home. I suppose it would be a search engine, in a way. 

#### Caveats and Further Thinking

The short of the long is that "traditional" foods do not really exist, and a more apt description is "regional foods" as borders and people change constantly. 

By "traditional" I mean recipies coming from the people either indigenous to a location or from cultures that currently inhabit the landscape. For example, the Québécois have not always lived in Québébec, being a product of French colonization in Canada, but there are traditionally Québécois foods like poutine and tourtière. Indigeneity to a region is not inherent to making a food traditional, though that certainly can be a major aspect of it. Another example of this is bannock. In many Indigenous communities of North America, bannock is considered a traditional food, but it dates to colonization and the import of wheat flour by Irish, English, and Scottish colonists. 

Furthermore, locations have not nor will always be stable. What we may consider stable, i.e. a nation-state, is not perpetual and everlasting. Foods claiming to originate in only one country may, instead, originate in that *region* but not necessarily because of a specific government. Borders do not contain culture. For this reason, several distinct countries located in the same geographic region may have similar foods, or have foods originating from colonization. An example of both is the empanada, which was initially brought to South America by Spanish colonists, but has since evolved to have regional differences between Souther American countries. 

As with cuisine, borders are not a fixed thing and change over time. While this LLM will likely have to utilize modern names for places, peoples, and dishes, there should also be a consideration for alternate mapping practices and historical populations. For this LLM, I would like to use region rather than country, but understand this makes life much more complicated. 

What could be done in the behind-the-scenes is a user can input any location they like, as specific as city to as broad as continent, using the place names they feel comfortable with. The algorithm then interprets that place name, "sees" where it exists on a map currently, and searches the internet (and user-submitted recipe database) to find recipes from that region. For example, a user could input "the Andes" as a location, the computer can interpret that the Andes region includes Chile, Bolivia, Ecuador, Columbia, and Argentina, and search for recipes based on that. 

Why do it this way? Because the internet is a current technology and does not necessarily contain all knowledge, especially specific regional knowledge. By placing an input region on a map, the computer can see what exists in that place presently, and search based on what exists there now, rather than reducing the search parameters by *only* looking up the specific region provided. 

I propose using the internet as a search function, simply because it would take a long time to accumulate enough user-submitted recipes to fill the whole globe. Ideally, this LLM would work with data that has specifically been input and consent has been obtained to utilize these recipes this way. Unfortunately, building a proper database based on user-submitted recipes would take quite a long time, and is prone to much more human error than if the LLM were to search the internet instead. At the beginning of this project (because I am thinking about it like a project), it could prioritize internet-sourced recipes, but once enough recipes have been submitted, they would become the primary source instead. 

### But First: What Is an LLM and How Does it Work? 

A large language model, or LLM, is a form of artificial intelligence (AI) model running on series of complex and interacting neural networks. They are classed as a form of generative AI (Pai, ch. 1, 2025), a type of AI model that creates an output of text or an image. This output is based on something called a "generative model" which means "training a model to produce new data that is similar to a given dataset." (Foster & Friston, 2023). This usually functions off predictive modelling and the neural network predicting the most likely next word in a sentence based on probability. Probablity is generated from examples of text, otherwise known as training data. With LLMs linked to the internet (as the all are, these days), they are privy to massive amounts of training data and can give intelligible responses on most topics. In saying so, LLMs are obviously subject to the biases present on the internet. If the training data is biased, the output will be, too. 

The exact specifics of how LLM neural networks work are beyond the scope of what is being asked of me in this task. The summary is that LLMs are trained on data, the same as all other AI models, but are trained on much, *much* more data. There are layers with differently weighted materials, and the weights are passed through layers and units to reach the most probable result (Foster & Friston, p. 26, 2023). It is, once again, probability based on large training datasets. 

#### In Context of This Assignment

For this project, the LLM would work more like a search function. The probability comes in when the model searches the web and finds the most likely dishes to be traditionally from a region, based on volume of results. Essentially, the model would automatically sift through search results for foods from a given region and evaluate which food is most likely to be traditional to the region based on the number of results, counting each dish then displaying, say, the top three most commonly found recipes. 

### Cultural Engagement and Inclusion

My idea is based around the ideas of cultural engagement and inclusion, but less so around augmented learning. The LLM cannot teach a user to cook, it is only for recipes. For cultural engagement, I tried not to focus on one culture, and instead wanted to help people explore cultures other than their own through food. As humans, we all have two things universally in common: we all have to eat and we are all going to die. Death and mourning cultures are something I have studied, but are not necessarily conducive to light, casual engagement. Alternatively, food does not require a person to die and can be enjoyed as often as a person is willing to cook it. 

Food can also tell us about the history of a place. Returning to the bannock example, this simple bread contains so much history for North American Indigenous peoples. Likewise, the empanada tells the story of Spanish colonization. The history of corn, even, is present in Mexican cuisine where it is transformed into all manner of food, kept whole, turned into dough, or left to develop fungus known as huitlacoche. Staple grain crops inform on the climate of a region and the history of agriculture therein. Places like Japan have a lot of fish in their cuisine because much of Japan's population is coastal. My point being, what people eat and where can say a lot about the culture of a given place. With my proposed tool being centred on food, it is also centred on culture. 

Inclusion was another consideration in creating an LLM idea. I did not want to centralize on any one culture, both because it would require more specificity in an LLM and because I wanted to make something that fostered engagement beyond oneself. When I think of cultural engagement, I tend to think on a global scale, rather than local. By making this a global project, everyone is able to participate. Additionally, there is an option for users to submit their own traditional/local/family recipes to the model and add them to the model's database. 

### Benefits, Limitations, and Ethical Concerns

The benefits of implementing this recipe-based LLM are not huge. I think it would be cool and potentially useful to some people, but it would not do something revolutionary. It could make life easier, sure, by compiling recipes for a user instead of you having to search yourself, but that labour is negligible. The only thing I believe is new is the user-submitted recipes, and that depends on the willingness of the audience. Most people, if they have a family recipe, they hang onto it, or, if they are particularly into cooking, will publish on a blog. Social media nowadays can fulfill the functions of a blog, and a whole website to post one recipe is not necessary. Again, this LLM would only compile recipes it found on the internet and from user-submitted items, not create its own. 

Even when considering the cultural learning and potential for inclusion, there are limitations in language. While LLMs can function as traslation platforms, that is not the primary function of my creative design. It *would* be useful to train this LLM to find recipes in different langauges, otherwise the range of what it can find would be limited to the English-speaking world, but doing so would require far more training data as well as involving people who speak other languages to ensure translations are accurate. If this were a long-term, well-funded project with globally interested parties, translation and sourcing would be easy. Unfortunately, this exercise is happening only in the theatre of the mind. 

Further limitations can be found in ethical concerns, specifically around data sourcing and envrionmental descrtuction. 

Data does not just fall out of a tree: it must be sourced from somewhere. In this case, the data used by my LLM would be sourced from users, which is not an issue as long as users are informed how their input will be used. Informed consent is the best-case-scenario for sourcing data. It is considered best practice in anthropological studies and clinical decisions, while AI companies hide how collected data will be used in dense privacy policies. Data privacy is an ongoing issue in AI fields (Zax, 2025). 

If the LLM runs only on consensually obtained user-input recipes, there is bias in language and therefor bias in which regions would be represented in the database. If I wanted to expand, I would have to source from more places. The LLM could run off internet searches, essentially. The blog, website, and digitized book authors have not consented to their work being used in this manner. Using book content can actually be considered piracy as books are solidly copyright-protected, as can be seen in the lawsuit against OpenAI in 2025 (Jamali, 2025). If I want to create an ethical LLM (and I remain doubtful this is possible), I want no part in my LLM committing piracy. The other option is to ask each individual blog, website, and author for permission to use their work for this project; in other words, a truly massive task. 

Environmentally, AI does not currently hold to a good standard. As AI companies and usage continue to grow, the data centres that host and run AI must also expand, with energy demand expected to hit 945 Terrawatt-hours by 2030 (International Energy Agency, p. 63).
